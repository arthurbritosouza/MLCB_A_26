```python
# ============================================================
# PROTÓTIPO NLU - Chatbot Agência de Viagens
# ============================================================
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# 1. Dataset próprio (15 frases, 4 intenções)
# ------------------------------------------------------------
dados_viagem = {
    'mensagem': [
        # comprar_passagem
        'Quero comprar uma passagem para o Rio de Janeiro',
        'Gostaria de reservar um voo para Lisboa',
        'Como faço para adquirir bilhetes aéreos?',
        'Preciso comprar passagens para duas pessoas',
        # cancelar_reserva
        'Quero cancelar minha reserva de hotel',
        'Como faço para cancelar meu voo?',
        'Preciso desmarcar a viagem que reservei',
        'Gostaria de cancelar a reserva número 12345',
        # falar_atendente
        'Quero falar com um atendente humano',
        'Pode me transferir para um consultor?',
        'Preciso de ajuda de uma pessoa de verdade',
        'Chame um atendente para mim, por favor',
        # consultar_pacote
        'Quais pacotes de viagem vocês tem para a praia?',
        'Me mostre as opções de pacotes para Europa',
        'Vocês tem pacote promocional para o Nordeste?',
    ],
    'intencao': [
        'comprar_passagem', 'comprar_passagem', 'comprar_passagem', 'comprar_passagem',
        'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva',
        'falar_atendente', 'falar_atendente', 'falar_atendente', 'falar_atendente',
        'consultar_pacote', 'consultar_pacote', 'consultar_pacote',
    ]
}

df_viagem = pd.DataFrame(dados_viagem)

# ------------------------------------------------------------
# 2. Divisão treino / teste
# ------------------------------------------------------------
X = df_viagem['mensagem']
y = df_viagem['intencao']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# ------------------------------------------------------------
# 3. Vetorização de texto (TF-IDF)
# ------------------------------------------------------------
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# ------------------------------------------------------------
# 4. Modelo de classificação (Regressão Logística)
# ------------------------------------------------------------
modelo_nlu = LogisticRegression(random_state=42)
modelo_nlu.fit(X_train_vec, y_train)

# Acurácia no conjunto de teste (holdout do próprio dataset)
acuracia_teste = modelo_nlu.score(X_test_vec, y_test)
print(f"Acurácia no conjunto de teste: {acuracia_teste * 100:.2f}%\n")

# ------------------------------------------------------------
# 5. Predição com frases INÉDITAS (fora do dataset de treino)
# ------------------------------------------------------------
frases_ineditas = [
    'Quero emitir uma passagem para Fortaleza',
    'Não quero mais minha reserva, pode cancelar',
    'Preciso urgente falar com um humano',
    'Tem pacote de viagem para Bariloche?',
]

frases_ineditas_vec = vectorizer.transform(frases_ineditas)
predicoes_ineditas = modelo_nlu.predict(frases_ineditas_vec)

# ------------------------------------------------------------
# 6. Exibição das intenções previstas
# ------------------------------------------------------------
print("=== Predições para frases inéditas ===")
for frase, intencao in zip(frases_ineditas, predicoes_ineditas):
    print(f"Frase: '{frase}'")
    print(f"  -> Intenção prevista: {intencao}\n")
```
